## 2. Array implementation of stack

```python
def prec(c):
    if c=="^":
        return 3
    elif c=="/" or c=="*":
        return 2
    elif c=="+" or c=="-":
        return 
    else:
        return -1
def isRightAssociative(c):
    return c==""
def infixToPostfix(s):
    st=[]
    res=[]
    for c in s:
        if ('a'<=c<='z')or('A'<=c<='Z')or('0'<=c<='9'):
            res.append(c)
        elif c=='(':
            st.append('(')
        elif c==')':
            while st and st[-1] !='(':
                res.append(st.pop())
            st.pop()
        else:
            while st and st[-1] != '(' and (
                prec(st[-1])>prec(c)or
                (prec(st[-1])== prec(c) and not isRightAssociative(c))
            ):
                res.append(st.pop())
            st.append(c)

    while st:
        res.append(st.pop())
    return''.join(res)
if __name__=='__main__':
    exp="a*(b+c)/d"
    print(infixToPostfix(exp))
        
```

## 3. Array implementation of Queue

```python
MAX=5
queue=[]
def enqueue():
    if len(queue)==MAX:
        print("Queue underflow")
    else:
        item=int(input("Enter element"))
        queue.append(item)
        print("Element inserted successfully")

def dequeue():
    if len(queue)==0:
        print("Queue underflow")
    else:
        print("Deleted element:",queue.pop(0))

def display():
    if len(queue)==0:
        print("Queue is empty")
    else:
        print("Queue elements:",queue)

while True:
    print("\n--QUEUE MENU--")
    print("1.enqueue")
    print("2.dequeue")
    print("3.display")
    print("4.exit")

    choice=int(input("Enter your choice"))
    if choice==1:
        enqueue()
    elif choice==2:
        dequeue()
    elif choice==3:
        display()
    elif choice==4:
        break
    else:
        print("Invalid choice")
```

## 4. Queue using linked list

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None

class Queue:
    def __init__(self):
        self.front=None
        self.rear=None

    def enqueue(self,data):
        new_node=Node(data)
        if self.rear is None:
            self.front=self.rear=new_node
        else:
            self.rear.next=new_node
            self.rear=new_node
            print("Element inserted successfully")

    def dequeue(self):
        if self.front is None:
            print("Queue underflow")
        else:
            data=self.front.data
            self.front=self.front.next
            if self.front is None:
                self.rear=None
                print("Deleted element:",data)
    def display(self):
        if self.front is None:
            print("Queue is empty")
        else:
            current=self.front
            print("Queue element:",end=" ")
            while current:
                print(current.data,end=" ")
                current=current.next
            print()

queue=Queue()
while True:
    print("\n--QUEUE MENU--")
    print("1.enqueue")
    print("2.dequeue")
    print("3.display")
    print("4.exit")

    choice=int(input("Enter your choice"))
    if choice==1:
        value=int(input("Enter an element"))
        queue.enqueue(value)
    elif choice==2:
        queue.dequeue()
    elif choice==3:
        queue.display()
    elif choice==4:
        break
    else:
        print("Invalid choice")
```

## 5. Binary Tree

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.left=None
        self.right=None
def inorder(root):
         if root:
            inorder(root.left)
            print(root.data,end=" ")
            inorder(root.right)
def preorder(root):
        if root:
            print(root.data,end=" ")
            preorder(root.left)
            preorder(root.right)
def postorder(root):
        if root:
            postorder(root.left)
            postorder(root.right)
            print(root.data,end=" ")
root=Node(1)
root.left=Node(2)
root.right=Node(3)
root.left.left=Node(4)
root.left.right=Node(5)
root.right.left=Node(6)
root.right.right=Node(7)
print("Inorder Traversal")
inorder(root)
print("\nPreorder Traversal")
preorder(root)
print("\nPostorder Traversal")
postorder(root)
```

## 6. Linear search Binary search

```python
def linear_search(arr,key):
    for i in range(len(arr)):
        if arr[i]==key:
            return i
            return -1

def binary_search(arr,key):
    low=0
    high=len(arr)-1
    while low<=high:
        mid=(low+high)//2
        if arr[mid]==key:
            return mid
        elif arr[mid]<key:
            low=mid+1
        else:
            high=mid-1
    return -1
arr=list(map(int,input("Enter elements seperate by space: ").split()))
key=int(input("Enter element to search: "))
position=linear_search(arr,key)
if position!=-1:
    print("Linear search element found at position: ",position)
else:
    print("Linear search: Element not found")
arr.sort()
print("Sorted array",arr)
position=binary_search(arr,key)
if position!=-1:
    print("Binary search: Element found at position: ",position)
else:
    print("Binary search: Element not found")
```
