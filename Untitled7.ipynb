```python
stack=[]
while True:
    print(f"\ncurrent stack:{stack}")
    print(f"Operations :\n1: Push \n2: Pop \n3: Peek \n4: Exit")
    choice=int(input("Enter choice (1-4):"))
    if choice==1:
        items=input("Enter value to push:")
        stack.append(items)
    elif choice==2:
        if stack:
            print(f"Popped:{stack.pop()}")
        else:
            print("Stack underflow empty stack")
    elif choice==3:
        if stack:
            print(f"Top element:{stack[-1]}")
        else:
            print("stack is empty")
    elif choice==4:
        print("End of subject")
        break
    else:
        print("Enter a valid input")
```

```python
def prec(c):
    if c=="^":
        return 3
    elif c=="/" or c=="*":
        return 2
    elif c=="+" or c=="-":
        return 
    else:
        return -1
def isRightAssociative(c):
    return c==""
def infixToPostfix(s):
    st=[]
    res=[]
    for c in s:
        if ('a'<=c<='z')or('A'<=c<='Z')or('0'<=c<='9'):
            res.append(c)
        elif c=='(':
            st.append('(')
        elif c==')':
            while st and st[-1] !='(':
                res.append(st.pop())
            st.pop()
        else:
            while st and st[-1] != '(' and (
                prec(st[-1])>prec(c)or
                (prec(st[-1])== prec(c) and not isRightAssociative(c))
            ):
                res.append(st.pop())
            st.append(c)

    while st:
        res.append(st.pop())
    return''.join(res)
if __name__=='__main__':
    exp="a*(b+c)/d"
    print(infixToPostfix(exp))
        
```
